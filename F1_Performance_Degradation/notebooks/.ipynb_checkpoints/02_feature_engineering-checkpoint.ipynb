```python
import sys
sys.path.append('../')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from src.feature_engineering import DegradationFeatureEngineer

# Load processed data from Notebook 1
df = pd.read_csv('../data/processed/races_2024_1-3.csv')

print(f"✅ Loaded {len(df)} laps")
print(f"📊 Shape: {df.shape}")
print(f"🏎️ Drivers: {df['Driver'].nunique()}")
print(f"🏁 Races: {df['EventName'].nunique()}")

df.head()
```

```python
# Create feature engineer
engineer = DegradationFeatureEngineer()

print("✅ Feature Engineer initialized")
print("🔧 Features to be created:")
print("  - Rolling statistics (mean, std, CV)")
print("  - Sector inconsistency scores")
print("  - Stint progression features")
print("  - Pace delta features")
print("  - Degradation labels")
```

```python
# Calculate rolling stats
df = engineer.calculate_rolling_stats(df, window=5)

print("✅ Rolling statistics calculated")
print("\n📊 New columns:")
print("  - LapTime_RollingMean")
print("  - LapTime_RollingStd (VARIANCE PROXY)")
print("  - LapTime_CV (Coefficient of Variation)")

# Show sample
df[['Driver', 'LapNumber', 'LapTime', 'LapTime_RollingMean', 
    'LapTime_RollingStd', 'LapTime_CV']].head(15)
```

```python
# Calculate sector inconsistency
df = engineer.calculate_sector_inconsistency(df)

print("✅ Sector inconsistency calculated")
print("\n📊 New columns:")
print("  - S1_Ratio, S2_Ratio, S3_Ratio")
print("  - S1_Ratio_Std, S2_Ratio_Std, S3_Ratio_Std")
print("  - SectorInconsistency_Score (combined)")

# Show sample
df[['Driver', 'LapNumber', 'SectorInconsistency_Score']].head(15)
```

```python
# Calculate stint features
df = engineer.calculate_stint_progression(df)

print("✅ Stint progression calculated")
print("\n📊 New columns:")
print("  - LapInStint")
print("  - StintProgress_Pct")
print("  - LapTimeDelta_FromStintStart")

# Show sample
df[['Driver', 'Stint', 'LapInStint', 'StintProgress_Pct', 
    'LapTimeDelta_FromStintStart']].head(15)
```

```python
# Calculate pace delta
df = engineer.calculate_pace_delta(df)

print("✅ Pace delta calculated")
print("\n📊 New columns:")
print("  - FastestLap_Race")
print("  - DeltaToFastest")
print("  - PersonalBest")
print("  - DeltaToPersonalBest")

# Show sample
df[['Driver', 'LapNumber', 'LapTime', 'PersonalBest', 
    'DeltaToPersonalBest']].head(15)
```

```python
# Create binary degradation label
df = engineer.create_degradation_label(df, threshold_std=0.5)

# Remove NaN rows
df = df.dropna().reset_index(drop=True)

print(f"✅ Degradation label created")
print(f"📊 Final shape: {df.shape}")
print(f"\n🔴 Degradation Statistics:")
print(f"  Total laps: {len(df)}")
print(f"  Degraded laps: {df['IsDegraded'].sum()}")
print(f"  Degraded %: {df['IsDegraded'].mean()*100:.2f}%")
print(f"  Normal laps: {(df['IsDegraded']==0).sum()}")

# Show distribution
df['IsDegraded'].value_counts()
```

```python
# Pick one driver and race for detailed analysis
driver = 'VER'
race = 1

driver_data = df[(df['Driver'] == driver) & (df['RaceNumber'] == race)].copy()

fig, axes = plt.subplots(3, 1, figsize=(15, 12))

# Plot 1: Lap time with degradation markers
axes[0].plot(driver_data['LapNumber'], driver_data['LapTime'], 
             marker='o', linewidth=2, markersize=5, label='Lap Time', color='blue')
degraded = driver_data[driver_data['IsDegraded'] == 1]
axes[0].scatter(degraded['LapNumber'], degraded['LapTime'], 
                color='red', s=150, label='Degraded Laps', zorder=5, marker='X')
axes[0].set_ylabel('Lap Time (s)', fontsize=12)
axes[0].set_title(f'{driver} - Lap Times with Degradation Detection (Race {race})', 
                  fontsize=14, fontweight='bold')
axes[0].legend(fontsize=11)
axes[0].grid(True, alpha=0.3)

# Plot 2: Rolling Std (Variance)
axes[1].plot(driver_data['LapNumber'], driver_data['LapTime_RollingStd'], 
             color='orange', marker='o', linewidth=2, markersize=5)
axes[1].axhline(y=0.5, color='red', linestyle='--', linewidth=2, 
                label='Degradation Threshold')
axes[1].fill_between(driver_data['LapNumber'], 0, driver_data['LapTime_RollingStd'], 
                      alpha=0.3, color='orange')
axes[1].set_ylabel('Lap Time Std Dev', fontsize=12)
axes[1].set_title('Performance Variance (5-lap rolling window)', fontsize=14, fontweight='bold')
axes[1].legend(fontsize=11)
axes[1].grid(True, alpha=0.3)

# Plot 3: Sector Inconsistency
axes[2].plot(driver_data['LapNumber'], driver_data['SectorInconsistency_Score'], 
             color='green', marker='o', linewidth=2, markersize=5)
axes[2].fill_between(driver_data['LapNumber'], 0, driver_data['SectorInconsistency_Score'], 
                      alpha=0.3, color='green')
axes[2].set_xlabel('Lap Number', fontsize=12)
axes[2].set_ylabel('Inconsistency Score', fontsize=12)
axes[2].set_title('Sector Timing Inconsistency', fontsize=14, fontweight='bold')
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('../results/figures/degradation_analysis.png', dpi=300, bbox_inches='tight')
plt.show()

print("💾 Figure saved to results/figures/degradation_analysis.png")
```
